```python
drop_cols = [
    "reviewerID",
    "first_review",
    "last_review",
    "1_star_reviews",
    "2_star_reviews",
    "3_star_reviews",
    "4_star_reviews",
    "5_star_reviews",
    "unique_products_reviewed"
]
df = df.drop(columns = drop_cols )
```

```python
df.columns = ['avg_rating', 'total_reviews', 'rating_variance', 'active_days',
       'reviews_per_day', 'percent_1Star', 'percent_2Star', 'percent_3Star', 'percent_4Star', 'percent_5Star',
       'total_helpful_votes', 'percent_helpful_per_review',
       'avg_days_between_reviews']
```

```python
features = df.columns

for col in features:
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    sns.histplot(df[col], kde=True)
    plt.title(f"Distribution of {col}")

    plt.subplot(1, 2, 2)
    sns.boxplot(x=df[col])
    plt.title(f"Boxplot of {col}")

    plt.tight_layout()
    plt.show()
```

>1. avg_rating -mostly people have positive interactions (i.e. 4-5 stars) with the products, a handful of people have low average ratings, which seems pretty normal.
>2. total_reviews - most of the people generally do not review many products or have just reviewed 2-3 products in their life, a very small number of people have reviewed hundreds of products. (These represent behavioral outliers and may be investigated further.)
>3. rating_variance - most of the people have variance of 0 which shows they rate consistently across products, while high variance shows varying ratings across different products.
>4. active_days - it's the difference between the date of the first and last review. This shows most people have given their first and last review on the same day, which is normal for people giving only one review or multiple reviews on the same day, while a few remain active through years.
>5. reviews_per_day - most have low review frequency (around 1-2 reviews per day or less) which is a normal behaviour, while I can see a few reviewers with very high review frequencies, which may contribute to anomaly detection.
>6. percen_1star to percent_4star - shows a clear picture of people rarely giving low ratings (1-4 stars).
>7. percent_5star - shows a clear picture of people frequently giving 5-star ratings, but a consistently high percentage of 5-star ratings from a single reviewer may contribute to suspicious behaviour when combined with other features.
>8. total_helpful_votes and percent_total_helpful_votes - most have near 0 helpful votes, but a few have very large numbers too. Higher helpful votes indicate that a reviewer's reviews are considered useful by other users, making this a useful behavioural feature.
>9. avg_days_between_reviews - the outliers are normal in this feature as most people review products after long intervals, while a few reviewers post reviews much more frequently.

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Reviewer Behaviour Features")
plt.show()
```

>1. avg_rating has a strong positive correlation with percent_5Star, indicating that reviewers with higher average ratings tend to give a larger proportion of 5-star reviews.
>2. avg_rating has a strong negative correlation with percent_1Star, showing that reviewers with higher average ratings rarely give 1-star reviews.
>3. reviews_per_day - It isn't showing very strong correlation with most other variables, which means it provides unique information.
>4. total_helpful_votes <--> percent_helpful_per_review (Moderately Strong Positive Correlation)<br>
>   -> total_helpful_votes captures overall reviewer influence.<br>
>   -> percent_helpful_per_review captures the average helpfulness of each review.

```python
df[["total_reviews", "reviews_per_day"]].corr()

plt.figure(figsize=(6,4))

sns.scatterplot(
    data=df,
    x="total_reviews",
    y="reviews_per_day",
    alpha=0.4
)
plt.title("Total Reviews vs Reviews per Day")
plt.xlabel("Total Reviews")
plt.ylabel("Reviews per Day")
plt.show()
```

> Most reviewers have very few total reviews and maintain a low review frequency. A few reviewers have unusually high review frequencies despite having a relatively small number of total reviews, while some reviewers have written hundreds of reviews over a long period with a low reviews-per-day ratio. These extreme points may represent behavioural outliers.

```python
plt.figure(figsize=(6,4))
sns.scatterplot(
    data=df,
    x="active_days",
    y="total_reviews",
    alpha=0.4
)

plt.title("Active Days vs Total Reviews")
plt.xlabel("Active Days")
plt.ylabel("Total Reviews")
plt.show()
```

> There is a moderate positive relationship between active days and total reviews, indicating that reviewers who remain active for longer periods generally contribute more reviews. However, there are exceptions where some reviewers have very high review counts within a relatively short active period, making them potential behavioural outliers.

```python
plt.figure(figsize=(6,4))
sns.scatterplot(
    data=df,
    x="avg_rating",
    y="rating_variance"
)

plt.title("Average Rating vs Rating Variance")
plt.xlabel("Average Rating")
plt.ylabel("Rating Variance")
plt.show()
```

> Reviewers with higher average ratings generally have lower rating variance, indicating more consistent rating behaviour. In contrast, reviewers with average ratings between 2 and 4 exhibit greater variance, showing a wider range of opinions across different products.

```python
plt.figure(figsize=(6,4))
sns.scatterplot(
    data=df,
    x="total_helpful_votes",
    y="total_reviews"
)

plt.title("Total Helpful Votes vs Total Reviews")
plt.xlabel("Total Helpful Votes")
plt.ylabel("Total Reviews")
plt.show()
```

> A weak positive relationship exists between helpful votes and total reviews. While reviewers with more reviews tend to receive more helpful votes, several reviewers with relatively few reviews have accumulated a high number of helpful votes, suggesting that review quality also influences helpfulness.

```python
plt.figure(figsize=(6,4))
sns.scatterplot(
    data=df,
    x="percent_5Star",
    y="avg_rating"
)

plt.title("% 5 Stars vs Average Rating")
plt.xlabel("% 5 Stars")
plt.ylabel("Average Rating")
plt.show()
```

> A strong positive relationship is observed between the percentage of 5-star ratings and the average rating. As the proportion of 5-star reviews increases, the reviewer's average rating also increases. This agrees with the strong positive correlation observed in the correlation heatmap.
