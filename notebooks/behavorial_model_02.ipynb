```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import MinMaxScaler
```

```python
df = pd.read_csv(r"E:\SGP Fake Review\Dataset\reviewer_dataset.csv")
```

```python
df.columns
```

```python
df_model = df[df['total_reviews']>=3]
```

```python
drop_cols = ["reviewerID",
             "first_review",
             "last_review",
             "1_star_reviews",
             "2_star_reviews",
             "3_star_reviews",
             "4_star_reviews",
             "5_star_reviews",
             "%_1Star",
             "%_2Star",
             "%_3Star",
             "%_4Star",
             "%_5Star",
             "unique_products_reviewed",
             "total_reviews",
             "active_days",
             "avg_rating",
             "%_helpful_per_review",
             "total_helpful_votes"]
df_model = df_model.drop(columns = drop_cols)
```

```python
df_model.reset_index(inplace = True, drop= True)
```

```python
df_model
```

```python
df_model['burst_intensity'] = df_model['review_burst_7d'] * df_model['burst_ratio_7d']
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_model)
```

```python
iso_01 = IsolationForest(
    n_estimators=200,
    contamination=0.01,
    random_state=42
)

_ = iso_01.fit(X_scaled)
print("Isolation Forest 01 trained successfully.")
```

```python
iso_02 = IsolationForest(
    n_estimators=200,
    contamination=0.02,
    random_state=42
)

_ = iso_02.fit(X_scaled)
print("Isolation Forest 02 trained successfully.")
```

```python
iso_03 = IsolationForest(
    n_estimators=200,
    contamination=0.03,
    random_state=42
)

_ = iso_03.fit(X_scaled)
print("Isolation Forest 03 trained successfully.")
```

```python
iso_04 = IsolationForest(
    n_estimators=200,
    contamination=0.04,
    random_state=42
)

_ = iso_04.fit(X_scaled)
print("Isolation Forest 04 trained successfully.")
```

```python
pred_iso_01 = iso_01.predict(X_scaled)
scores_iso_01 = iso_01.decision_function(X_scaled)

score_scaler = MinMaxScaler((0,100))

behavior_score_iso_01 = score_scaler.fit_transform(
    scores_iso_01.reshape(-1,1)
)
```

```python
pred_iso_02 = iso_02.predict(X_scaled)
scores_iso_02 = iso_02.decision_function(X_scaled)

score_scaler = MinMaxScaler((0,100))

behavior_score_iso_02 = score_scaler.fit_transform(
    scores_iso_02.reshape(-1,1)
)
```

```python
pred_iso_03 = iso_03.predict(X_scaled)
scores_iso_03 = iso_03.decision_function(X_scaled)

score_scaler = MinMaxScaler((0,100))

behavior_score_iso_03 = score_scaler.fit_transform(
    scores_iso_03.reshape(-1,1)
)
```

```python
pred_iso_04 = iso_04.predict(X_scaled)
scores_iso_04 = iso_04.decision_function(X_scaled)

score_scaler = MinMaxScaler((0,100))

behavior_score_iso_04 = score_scaler.fit_transform(
    scores_iso_04.reshape(-1,1)
)
```

```python
df_model["prediction_01"] = pred_iso_01

df_model["decision_score_01"] = scores_iso_01

df_model["behavior_score_01"] = behavior_score_iso_01
```

```python
df_model["prediction_02"] = pred_iso_02

df_model["decision_score_02"] = scores_iso_02

df_model["behavior_score_02"] = behavior_score_iso_02
```

```python
df_model["prediction_03"] = pred_iso_03

df_model["decision_score_03"] = scores_iso_03

df_model["behavior_score_03"] = behavior_score_iso_03
```

```python
df_model["prediction_04"] = pred_iso_04

df_model["decision_score_04"] = scores_iso_04

df_model["behavior_score_04"] = behavior_score_iso_04
```

```python
df_model
```

```python
contamination_values = [0.01, 0.02, 0.03, 0.04]

for c in contamination_values:
    model = IsolationForest(
        contamination=c,
        random_state=42
    )

    model.fit(X_scaled)

    pred = model.predict(X_scaled)
    n_anomalies = (pred == -1).sum()

    print(f"Contamination: {c}")
    print(f"Anomalies detected: {n_anomalies}")
    print("-" * 40)
```

```python
df_model_sort = df_model.sort_values(by= 'behavior_score_01')
```

```python
df_model_sort.reset_index(inplace = True)
```
