```python
df.drop(columns = ['title', 'text', 'images', 'parent_asin'], inplace = True)
```

```python
df.set_index('user_id', inplace = True)
```

```python
df.reset_index(inplace = True)
```

```python
df.rename(columns = {"user_id":"reviewerID"}, inplace =True)
```

```python
df['review_datetime'] = pd.to_datetime(df['timestamp'], unit= 'ms')
```

```python
df.head()
```

```python
df["review_datetime"].head(25).dt.strftime("%Y-%m-%d %H:%M:%S")
```

```python
df_reviewer = df.groupby('reviewerID')['rating'].agg(
    avg_rating = 'mean',
    total_reviews = 'count',
    rating_variance = 'var',
    
)
```

```python
rating_entropy = (
    df.groupby('reviewerID')['rating']
           .apply(lambda x: entropy(x.value_counts(normalize=True), base=2))
           .rename('rating_entropy')
)
df_reviewer['rating_entropy'] = rating_entropy
```

```python
df_reviewer.head()
```

```python
reviewer_dates = df.groupby("reviewerID")["review_datetime"].agg(
    first_review="min",
    last_review="max"
)
```

```python
reviewer_dates.head()
```

```python
reviewer_dates['active_days'] = (reviewer_dates['last_review']- reviewer_dates['first_review']).dt.days + 1
```

```python
df_reviewer = df_reviewer.join(reviewer_dates)
```

```python
df_reviewer.reset_index(inplace =  True)
```

```python
df_reviewer.head()
```

```python
df_reviewer['rating_variance'] = df_reviewer['rating_variance'].fillna(0)
```

```python
df_reviewer
```

```python
df_reviewer['reviews/day'] = df_reviewer['total_reviews']/df_reviewer['active_days']
```

```python

def max_reviews_in_window(group, window='7D'):
    dates = group.sort_values()

    s = pd.Series(1, index=dates)

    return int(s.rolling(window).sum().max())

burst_7d = (
    df.groupby('reviewerID')['review_datetime']
           .apply(lambda x: max_reviews_in_window(x, '7D'))
           .rename('max_reviews_7d')
)
df_reviewer = df_reviewer.merge(
    burst_7d,
    on='reviewerID',
    how='inner'
)
```

```python
df_reviewer.rename(
    columns={'max_reviews_7d': 'review_burst_7d'},
    inplace=True
)
```

```python
df_reviewer['burst_ratio_7d'] = (
    df_reviewer['review_burst_7d'] /
    df_reviewer['total_reviews']
)
```

```python
df_reviewer[(df_reviewer['burst_ratio_7d'] >=0.8) & (df_reviewer['active_days']>=7)]
```

```python
for rating in range(1, 6):
    rating_count = (
        df[df["rating"] == rating]
        .groupby("reviewerID")
        .size()
    )

    df_reviewer[f"{rating}_star_reviews"] = (
        df_reviewer["reviewerID"]
        .map(rating_count)
        .fillna(0)
        .astype(int)
    )

    df_reviewer[f"%_{rating}Star"] = (
        df_reviewer[f"{rating}_star_reviews"]
        / df_reviewer["total_reviews"]
    ) * 100
```

```python
df_reviewer.head()
```

```python
unique_products = df.groupby("reviewerID")["asin"].nunique()

df_reviewer["unique_products_reviewed"] = (
    df_reviewer["reviewerID"]
    .map(unique_products)
)
```

```python
helpful_votes = df.groupby("reviewerID")["helpful_vote"].sum()

df_reviewer["total_helpful_votes"] = (
    df_reviewer["reviewerID"]
    .map(helpful_votes)
    .fillna(0)
    .astype(int)
)
```

```python
df_reviewer["%_helpful_per_review"] = df_reviewer["total_helpful_votes"]/ df_reviewer["total_reviews"]
```

```python
df_sort = df.sort_values(["reviewerID", "review_datetime"])
```

```python
df_sort["days_between_reviews"] = (
    df_sort.groupby("reviewerID")["review_datetime"]
      .diff()
      .dt.days
)
```

```python
avg_gap = (
    df_sort.groupby("reviewerID")["days_between_reviews"]
      .mean()
)
df_reviewer["avg_days_between_reviews"] = (
    df_reviewer["reviewerID"]
      .map(avg_gap)
      .fillna(0)
)
```
